```python
import matplotlib.pyplot as plt
x = [10,15,20,25,30,35,40,45,50]
y = [5000,2000,3000,4000,5000,6000,7000,8000,9000]
plt.plot(x,y)                   # line graphs
plt.title("My First Plot")
plt.xlabel("X axes")
plt.ylabel("Y axes")
plt.show()                      # this is compulsary to sow the data 
```

```python
plt.bar(x,y)        # this is print Bar graph 
```

```python
plt.scatter(x,y)            #this print the point 
```

```python
plt.figure(figsize=(8,5))
plt.plot(x,y ,color ='red',marker = 'o',linestyle = '--',linewidth = 2 , markersize = 8)
plt.title("Comparison of age Vs salary")
plt.xlabel("Age")
plt.ylabel("Salary")
plt.show()
```
